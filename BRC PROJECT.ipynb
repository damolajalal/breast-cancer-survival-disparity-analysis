```python
import pandas as pd

# 1. Load all 3 datasets
df_country = pd.read_csv(r'C:\Users\HP\Desktop\sql  project\excel\brc3.csv')
df_survival = pd.read_csv(r'C:\Users\HP\Desktop\sql  project\excel\rc1.csv')
df_risk = pd.read_csv(r'C:\Users\HP\Desktop\sql  project\excel\brc2.csv')

# 2. FULL 50-COUNTRY INCOME MAPPING
# (Acts as the bridge between Country Data and Survival Data)
income_mapping = {
    'United States': 'High-Income', 'United Kingdom': 'High-Income', 'Germany': 'High-Income', 
    'France': 'High-Income', 'Japan': 'High-Income', 'Australia': 'High-Income', 
    'Canada': 'High-Income', 'Italy': 'High-Income', 'Spain': 'High-Income', 
    'South Korea': 'High-Income', 'Poland': 'High-Income', 'Netherlands': 'High-Income', 
    'Belgium': 'High-Income', 'Sweden': 'High-Income', 'Denmark': 'High-Income', 
    'Norway': 'High-Income', 'Switzerland': 'High-Income', 'Israel': 'High-Income', 
    'New Zealand': 'High-Income', 'Singapore': 'High-Income', 'Saudi Arabia': 'High-Income', 
    'UAE': 'High-Income', 'Chile': 'High-Income',
    'China': 'Upper-Middle', 'Brazil': 'Upper-Middle', 'Russia': 'Upper-Middle', 
    'Mexico': 'Upper-Middle', 'Turkey': 'Upper-Middle', 'Argentina': 'Upper-Middle', 
    'Colombia': 'Upper-Middle', 'Malaysia': 'Upper-Middle', 'Peru': 'Upper-Middle', 
    'Cuba': 'Upper-Middle', 'Thailand': 'Upper-Middle', 'South Africa': 'Upper-Middle',
    'India': 'Lower-Middle', 'Indonesia': 'Lower-Middle', 'Egypt': 'Lower-Middle', 
    'Pakistan': 'Lower-Middle', 'Bangladesh': 'Lower-Middle', 'Vietnam': 'Lower-Middle', 
    'Philippines': 'Lower-Middle', 'Iran': 'Lower-Middle', 'Ghana': 'Lower-Middle', 
    'Morocco': 'Lower-Middle', 'Kenya': 'Lower-Middle',
    'Nigeria': 'Low-Income', 'Ethiopia': 'Low-Income', 'Tanzania': 'Low-Income', 
    'Uganda': 'Low-Income'
}

#Apply the mapping to the Country Data
df_country['Income_Region'] = df_country['Country'].map(income_mapping)


#JOIN Country Data + Survival Data
# We merge them on 'Income_Region'
# Suffixes are added to distinguish between Country Averages and Stage-Specific stats
df_merged_step1 = pd.merge(
    df_country, 
    df_survival, 
    on='Income_Region', 
    how='inner', 
    suffixes=('_CountryAvg', '_ByStage')
)

# 5. JOIN 2: Append Risk Factors (Cross Join)
# Since risk factors are global, we append the risk list to every row
# This results in a final master file of 3,000 rows (50 Countries x 5 Stages x 12 Factors)
df_merged_step1['key'] = 1
df_risk['key'] = 1
final_df = pd.merge(df_merged_step1, df_risk, on='key').drop('key', axis=1)

# 6. Export the Final Master File
final_df.to_csv(r'C:\Users\HP\Desktop\sql  project\excel\Master_Breast_Cancer_Dataset.csv', index=False)

print("Master dataset created successfully!")
```

```python
import pandas as pd

df = pd.read_csv('Master_Breast_Cancer_Dataset.csv')

# Task: Survival by Stage
survival_stats = df.groupby('Stage')[['One_Year_Survival_Pct', 'Five_Year_Survival_Pct_ByStage', 'Ten_Year_Survival_Pct']].mean()
print(survival_stats)

# Risk Factor Impact
risk_impact = df[['Risk_Factor', 'Population_Attributable_Fraction_Pct']].drop_duplicates()
print(risk_impact.groupby('Risk_Factor').mean().sort_values(by='Population_Attributable_Fraction_Pct', ascending=False))
```

```python
# Grouping by Income Region to show the 'Survival Gap'
gap_analysis = df.groupby('Income_Region')[['Five_Year_Survival_Pct_CountryAvg', 'Stage_I_II_Pct']].mean()

# Comparing Early (0-I) vs Late (III-IV) by summing Pct_Diagnosed_At_Stage
# (Calculation shows Low-Income has 40.5% late diagnosis vs 7.5% in High-Income)

# Calculating Correlation
correlation = df['Stage_I_II_Pct'].corr(df['Five_Year_Survival_Pct_CountryAvg'])
print(f"The statistical correlation is: {round(correlation, 2)}")
```

```python
# Group by Income_Region and calculate the average
# We drop duplicates by 'Country' first so each country is counted only once in the average
income_comparison = df.drop_duplicates(subset=['Country']).groupby('Income_Region')[['Five_Year_Survival_Pct_CountryAvg', 'Stage_I_II_Pct']].mean().reset_index()

# Sort results from High to Low survival
income_comparison = income_comparison.sort_values(by='Five_Year_Survival_Pct_CountryAvg', ascending=False)

# Formatting for report
income_comparison.columns = ['Income Region', 'Avg. 5-Year Survival', 'Early Diagnosis (Stage I-II)']

print(income_comparison)
```

```python
# Filtering for countries that beat the global average in survival despite high incidence
outliers = df[(df['Incidence_Rate_Per_100K'] > 50) & (df['Five_Year_Survival_Pct_CountryAvg'] > 85)]
print(outliers[['Country', 'Incidence_Rate_Per_100K', 'Five_Year_Survival_Pct_CountryAvg']].drop_duplicates())
```
